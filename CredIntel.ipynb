```python

import pandas as pd

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00350/default%20of%20credit%20card%20clients.xls"

df = pd.read_excel(url, header=1)

print("Dataset downloaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

df.head()
```

```python

# 1. Check the dataset size
print("Dataset shape:", df.shape)

# 2. Inspect column names
print("\nColumn names:")
print(df.columns.tolist())

# 3. Check for missing values
print("\nMissing values per column:")
print(df.isnull().sum())

# 4. Check for duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# 5. Inspect the default outcome
print("\nDefault outcome counts:")
print(df["default payment next month"].value_counts())
```

```python

# Display the first 10 column names and their data types
column_info = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values
})

display(column_info)

# Show basic statistics for numeric columns
display(df.describe().round(2))
```

```python

import sqlite3

# Connect to an in-memory SQLite database
conn = sqlite3.connect(":memory:")

# Copy our real dataset into a SQL table
df.to_sql("credit_clients", conn, if_exists="replace", index=False)

# Check that SQL can access our data
result = pd.read_sql_query("""
    SELECT COUNT(*) AS total_customers
    FROM credit_clients;
""", conn)

display(result)
```

```python

query = """
SELECT
    COUNT(*) AS total_customers,
    SUM("default payment next month") AS customers_defaulted,
    ROUND(
        100.0 * SUM("default payment next month") / COUNT(*),
        2
    ) AS default_rate_percent
FROM credit_clients;
"""

default_summary = pd.read_sql_query(query, conn)
display(default_summary)
```

```python

query = """
SELECT
    PAY_0 AS repayment_status,
    COUNT(*) AS total_customers,
    SUM("default payment next month") AS customers_defaulted,
    ROUND(
        100.0 * SUM("default payment next month") / COUNT(*),
        2
    ) AS default_rate_percent
FROM credit_clients
GROUP BY PAY_0
ORDER BY default_rate_percent DESC;
"""

repayment_analysis = pd.read_sql_query(query, conn)
display(repayment_analysis)
```

```python

import plotly.express as px

fig = px.bar(
    repayment_analysis.sort_values("repayment_status"),
    x="repayment_status",
    y="default_rate_percent",
    hover_data=["total_customers", "customers_defaulted"],
    title="Credit Card Default Rate by Repayment Status",
    labels={
        "repayment_status": "Repayment Status (PAY_0)",
        "default_rate_percent": "Default Rate (%)"
    },
    text="default_rate_percent"
)

fig.update_traces(texttemplate="%{text:.2f}%", textposition="outside")
fig.update_layout(
    xaxis=dict(type="category"),
    yaxis_title="Default Rate (%)",
    xaxis_title="Repayment Status",
    template="plotly_white"
)

fig.show()
```

```python

query = """
SELECT
    CASE
        WHEN LIMIT_BAL < 50000 THEN 'Under 50K'
        WHEN LIMIT_BAL < 100000 THEN '50K–99K'
        WHEN LIMIT_BAL < 200000 THEN '100K–199K'
        WHEN LIMIT_BAL < 500000 THEN '200K–499K'
        ELSE '500K and above'
    END AS credit_limit_segment,

    COUNT(*) AS total_customers,

    SUM("default payment next month") AS customers_defaulted,

    ROUND(
        100.0 * SUM("default payment next month") / COUNT(*),
        2
    ) AS default_rate_percent

FROM credit_clients
GROUP BY credit_limit_segment
ORDER BY
    CASE credit_limit_segment
        WHEN 'Under 50K' THEN 1
        WHEN '50K–99K' THEN 2
        WHEN '100K–199K' THEN 3
        WHEN '200K–499K' THEN 4
        ELSE 5
    END;
"""

credit_limit_analysis = pd.read_sql_query(query, conn)
display(credit_limit_analysis)
```

# looks great

```python

fig = px.bar(
    credit_limit_analysis,
    x="credit_limit_segment",
    y="default_rate_percent",
    hover_data=["total_customers", "customers_defaulted"],
    title="Credit Card Default Rate by Credit Limit",
    labels={
        "credit_limit_segment": "Credit Limit Segment (NTD)",
        "default_rate_percent": "Default Rate (%)"
    },
    text="default_rate_percent"
)

fig.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside"
)

fig.update_layout(
    xaxis_title="Credit Limit Segment",
    yaxis_title="Observed Default Rate (%)",
    template="plotly_white"
)

fig.show()
```

```python

def credintel_analyst(sql_query):
    """
    Run a read-only SQL query against CredIntel's database
    and return the actual results.
    """
    cleaned_query = sql_query.strip().lower()

    if not cleaned_query.startswith("select"):
        raise ValueError("For safety, only SELECT queries are allowed.")

    if ";" in cleaned_query.rstrip(";"):
        raise ValueError("Please run one SQL query at a time.")

    result = pd.read_sql_query(sql_query, conn)
    return result


# Test the analyst with a real banking question
answer = credintel_analyst("""
    SELECT
        COUNT(*) AS total_customers,
        SUM("default payment next month") AS customers_defaulted,
        ROUND(
            100.0 * SUM("default payment next month") / COUNT(*),
            2
        ) AS default_rate_percent
    FROM credit_clients
""")

display(answer)
```

```python

def ask_credintel(question):
    question_lower = question.lower()

    if "default rate" in question_lower or "default percentage" in question_lower:
        sql = """
        SELECT
            COUNT(*) AS total_customers,
            SUM("default payment next month") AS customers_defaulted,
            ROUND(
                100.0 * SUM("default payment next month") / COUNT(*), 2
            ) AS default_rate_percent
        FROM credit_clients
        """
        result = credintel_analyst(sql)
        row = result.iloc[0]

        return (
            f"CredIntel analyzed {row['total_customers']:,} customers. "
            f"{row['customers_defaulted']:,} defaulted in the recorded "
            f"outcome period. The observed default rate was "
            f"{row['default_rate_percent']:.2f}%."
        )

    elif "credit limit" in question_lower:
        return credit_limit_analysis.to_string(index=False)

    elif "repayment" in question_lower:
        return repayment_analysis.to_string(index=False)

    else:
        return (
            "I can currently answer questions about the overall default rate, "
            "credit-limit segments, and repayment-status groups. "
            "Try asking about one of those topics."
        )


# Test CredIntel
print(ask_credintel("What is the default rate?"))
```

```python

def credintel_ai(question):
    # Retrieve real statistics from the SQLite database
    sql = """
    SELECT
        COUNT(*) AS total_customers,
        SUM("default payment next month") AS customers_defaulted,
        ROUND(
            100.0 * SUM("default payment next month") / COUNT(*), 2
        ) AS default_rate_percent
    FROM credit_clients
    """

    actual_results = credintel_analyst(sql).to_dict(orient="records")

    prompt = f"""
You are CredIntel, a careful banking analytics assistant.

Answer the user's question using the verified database results below.

Rules:
- Treat the supplied results as the source of truth.
- Do not invent statistics or claim unobserved findings.
- Distinguish correlation from causation.
- Explain that this is historical data, not current lending guidance.
- If the results don't answer the question, say so.

Verified SQL results:
{actual_results}

User question:
{question}
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )

    return response.text


print(credintel_ai(
    "What is the overall default rate, and what does it mean?"
))
```

```python

def credintel_repayment_ai(question):
    # Fetch actual repayment-status statistics from SQLite
    sql = """
    SELECT
        PAY_0 AS repayment_status,
        COUNT(*) AS total_customers,
        SUM("default payment next month") AS customers_defaulted,
        ROUND(
            100.0 * SUM("default payment next month") / COUNT(*), 2
        ) AS default_rate_percent
    FROM credit_clients
    GROUP BY PAY_0
    ORDER BY PAY_0
    """

    results = credintel_analyst(sql).to_dict(orient="records")

    prompt = f"""
You are CredIntel, a banking analytics assistant.

Answer the user's question using only the supplied SQL results.

Rules:
- Use actual values from the results.
- Do not invent figures or unsupported explanations.
- Consider group sizes when comparing default rates.
- Explain that observed associations do not establish causation.
- The dataset is historical, not a current lending recommendation.
- If the question cannot be answered from these results, say so.

Verified SQL results:
{results}

User question:
{question}
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )

    return response.text


print(credintel_repayment_ai(
    "Which repayment statuses have the highest observed default rates?"
))
```
